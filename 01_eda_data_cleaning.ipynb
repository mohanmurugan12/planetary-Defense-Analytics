```python
# Load dataset directly from data folder
df = pd.read_csv('data/pha_dataset.csv')

# Check dataset shape
print("Dataset Shape:", df.shape)

# Display first 5 rows
df.head()
```

```python
# Check columns and data types
df.info()
```

```python
# Check missing values for all columns
missing_data = df.isnull().sum()
missing_data_percent = (df.isnull().sum() / len(df)) * 100

# Combine count and percentage
missing_df = pd.DataFrame({
    'Missing Values': missing_data,
    'Percentage (%)': missing_data_percent
})

# Display columns that have missing values
missing_df[missing_df['Missing Values'] > 0]
```

```python
# Check target variable distribution for Case 1
print(df['pha'].value_counts(dropna=False))
print("\nPercentage Share:")
print(df['pha'].value_counts(normalize=True, dropna=False) * 100)
```

```python
# Drop rows where 'pha' is NaN
df = df.dropna(subset=['pha']).copy()

# Map 'Y' to 1 and 'N' to 0
df['pha'] = df['pha'].map({'Y': 1, 'N': 0}).astype(int)

# Verify clean target distribution
print("Cleaned Target Distribution:")
print(df['pha'].value_counts())
print("\nNew Dataset Shape:", df.shape)
```

```python
# Drop ID columns and columns with > 50% missing values
cols_to_drop = ['id', 'spkid', 'full_name', 'pdes', 'name', 'prefix', 'orbit_id', 'equinox']

# Calculate missing percentage
missing_percent = (df.isnull().sum() / len(df)) * 100
high_missing_cols = missing_percent[missing_percent > 50].index.tolist()

# Combine all columns to drop
all_drop_cols = list(set(cols_to_drop + high_missing_cols))

# Drop columns
df_cleaned = df.drop(columns=[col for col in all_drop_cols if col in df.columns])

print("Remaining Columns Count:", df_cleaned.shape[1])
print("Cleaned Dataset Shape:", df_cleaned.shape)
```

```python
# Identify numerical columns
num_cols = df_cleaned.select_dtypes(include=['float64', 'int64']).columns

# Fill missing values in numerical columns with Median
for col in num_cols:
    if df_cleaned[col].isnull().sum() > 0:
        df_cleaned[col] = df_cleaned[col].fillna(df_cleaned[col].median())

# Verify if any missing values remain
print("Remaining Missing Values Count:", df_cleaned.isnull().sum().sum())
```

```python
# Save cleaned dataset to data folder
df_cleaned.to_csv('data/cleaned_pha_dataset.csv', index=False)

print("Cleaned dataset saved successfully as 'data/cleaned_pha_dataset.csv'!")
```

```python
# Fill remaining missing values in categorical columns with Mode
cat_cols = df_cleaned.select_dtypes(include=['object']).columns
for col in cat_cols:
    if df_cleaned[col].isnull().sum() > 0:
        df_cleaned[col] = df_cleaned[col].fillna(df_cleaned[col].mode()[0])

# Final check for missing values
print("Final Missing Values Count:", df_cleaned.isnull().sum().sum())

# Re-save cleaned dataset
df_cleaned.to_csv('data/cleaned_pha_dataset.csv', index=False)
print("Updated cleaned dataset saved successfully!")
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Plot 1: Target Variable Distribution (pha)
plt.figure(figsize=(6, 4))
sns.countplot(x='pha', data=df_cleaned)
plt.title('Distribution of Potentially Hazardous Asteroids (PHA)')
plt.xlabel('PHA (0 = Non-Hazardous, 1 = Hazardous)')
plt.ylabel('Count')
plt.show()
```

```python
# Automatically select numerical columns present in df_cleaned
num_cols = df_cleaned.select_dtypes(include=['float64', 'int64']).columns

# Filter top features to avoid overcrowded heatmap
sample_cols = [col for col in ['a', 'e', 'i', 'om', 'w', 'q', 'ad', 'H', 'pha'] if col in num_cols]

plt.figure(figsize=(10, 8))
sns.heatmap(df_cleaned[sample_cols].corr(), annot=True, fmt='.2f', cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Heatmap of Available Key Features')
plt.show()
```

```python
# Plot 3: Distribution of Absolute Magnitude (H)
plt.figure(figsize=(8, 5))
sns.histplot(df_cleaned['H'], bins=50, kde=True, color='purple')
plt.title('Distribution of Absolute Magnitude (H)')
plt.xlabel('Absolute Magnitude (H)')
plt.ylabel('Frequency')
plt.show()
```
